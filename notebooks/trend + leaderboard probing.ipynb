```python
# ======================
# 1. SETUP & DATA LOADING
# ======================
from google.colab import userdata
import os
import pandas as pd
from tqdm import tqdm
import kagglehub

os.environ['KAGGLE_USERNAME'] = userdata.get('KAGGLE_USERNAME')
os.environ['KAGGLE_KEY'] = userdata.get('KAGGLE_KEY')

path = kagglehub.competition_download('recommender-system-challenge-2526-s-2')
print(f" Data loaded from: {path}")

sample_submission_df = pd.read_csv(f'{path}/sample_submission.csv')
train_df = pd.read_csv(f'{path}/train.csv')
test_df = pd.read_csv(f'{path}/test.csv')

print(" Data loaded successfully!")

```

```python
full_df = pd.concat([train_df, test_df])

cutoff_time = full_df['timestamp'].quantile(0.9)
recent_df = full_df[full_df['timestamp'] >= cutoff_time]

recent_popular_items = recent_df['item_id'].value_counts().index.tolist()


global_popular_items = full_df['item_id'].value_counts().index.tolist()


user_histories = full_df.groupby('user_id')['item_id'].apply(set).to_dict()

target_users = sample_submission_df['user_id'].values
predictions = []


for user_id in tqdm(target_users, desc="Submission generation"):
    seen_items = user_histories.get(user_id, set())

    top_10 = []


    for item in recent_popular_items:
        if item not in seen_items:
          top_10.append(str(item))
        if len(top_10) == 10:
            break

    #Fallback
    if len(top_10) < 10:
        for item in global_popular_items:
            if item not in seen_items and str(item) not in top_10:
                top_10.append(str(item))
            if len(top_10) == 10:
                break

    predictions.append(",".join(top_10))


```

```python

submission_df = sample_submission_df.copy()
submission_df['item_id'] = predictions
print(submission_df['item_id'][0])

submission_df.to_csv('submission_recent_trends.csv', index=False)
print("submission_recent_trends.csv generated successfully! Ready to submit.")
```
